### if-else statements

```python
x = input()  # x will equal "3", since input() defaults to string
if x == 3:  # WARNING! This does not do what you expect it to - the types have to be the same too
    print("Correct")
else:
    print("Incorrect")
```

```python
x = int(input())  # this is called casting - changing one data type to another
# str() changes it to a string
if x == 3:  # now this does what you expect it to
    print("Correct")
else:
    print("Incorrect")
```

```python
x = input("What food do you like")  # you can put a string in input
if x == "Waffles" or x == "Pasta":  # case matters: "Waffles" and "waffles" are NOT the same
    print("Yum")
else:
    print("Yuck")
```

```python
x = input("What's your favorite color")
y = int(input("What's your favorite number"))
if x == "Blue" and y == 3:
    print("Yay")
else:
    print("NOOOOOOOOOO")
```

**Booleans**

These are another data type.
They can only be two values: True or False

```python
if True:
    print("True")  # this wall always print
if False:
    print("False")  # this will never print
```

```python
x = int(input("What's your favorite number")) == 42  # x will evaluate to True or False
if x:  # this is the same as 'if x == True'
    print("Yay")
else:
    print("Nooooo")  # This will execute if x is False
```

```python
x = (2 == 2)  # evaluates to True
print(x)
x = 2 == 2  # parentheses not required, statements on the right of '=' are always evaluated first 
print(x)
```

```python
y = 2 == 3  # evaluates to False
print(x, y)
```

### Loops

Loops allow you to execute statements multiple times:<br>
The syntax is:<br>
for *variable* in range(*begin*, *end*, *increment*):<br>
&emsp;statements

```python
for i in range(0, 3, 1):  # this goes from 0-2, incrementing by 1
    print(i)  # note that the end value - 3 - is excluded
```

```python
# only using 2 parameters defaults increment to 1
for i in range(2, 6):  # this goes from 2-5, incrementing by 1
    print(i)
```

```python
# only using 1 parameter defaults begin to 0, defaults increment to 1
for i in range(5):  # this goes from 0-4, incrementing by 1
    print(i)
```

```python
arr1 = [1, 2, 3, 4, 5]
for i in range(len(arr1)):  # prints every single element in the list
    print(arr1[i])
```

```python
for element in arr1:  # this is called a for-each loop - it goes through every single element of arr1
    print(element)
```

```python
# arrays in arrays! (2D arrays)
arr1 = [[1, 2, 3], [4, 5, 6], ["bob", "says", "hi"], ["i", "like", "potatoes"]]
for outer in arr1:
    for inner in outer:
        print(inner, end=' ')  # prints inner + a space
    print()  # prints newline character
```

```python
# Sometimes 2D arrays are represented this way:
# this only changes readability, not functionality
arr1 = [
        [1, 2, 3],
        [4, 5, 6],
        ["bob", "says", "hi"],
        ["i", "like", "potatoes"]
       ]
for outer in arr1:
    for inner in outer:
        print(inner, end=' ')  # prints inner + a space
    print()  # prints newline character
```
